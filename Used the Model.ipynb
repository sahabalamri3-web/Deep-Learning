```python
import gradio as gr
import cv2
import numpy as np
from ultralytics import YOLO
from PIL import Image

# مسار المودل
MODEL_PATH = "/content/drive/MyDrive/Colab Notebooks/best.pt"
model = YOLO(MODEL_PATH)

def detect_image(image):
    results = model(image)
    annotated = results[0].plot()
    return annotated

def detect_video(video):
    cap = cv2.VideoCapture(video)
    frames = []

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        results = model(frame)
        annotated = results[0].plot()
        frames.append(cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB))

    cap.release()
    return frames

with gr.Blocks(title="Vehicle Detection - YOLOv8") as demo:
    gr.Markdown("## 🚗 Vehicle Detection using YOLOv8")

    with gr.Tab("Image"):
        img_in = gr.Image(type="pil", label="Upload Image")
        img_out = gr.Image(label="Detected Image")
        btn_img = gr.Button("Detect")
        btn_img.click(detect_image, inputs=img_in, outputs=img_out)

    with gr.Tab("Video"):
        vid_in = gr.Video(label="Upload Video")
        vid_out = gr.Gallery(label="Detected Frames", columns=3)
        btn_vid = gr.Button("Detect")
        btn_vid.click(detect_video, inputs=vid_in, outputs=vid_out)

demo.launch(share=True)
```

```python
import gradio as gr
import cv2
import os
from ultralytics import YOLO

# مسار المودل
MODEL_PATH = "/content/drive/MyDrive/Colab Notebooks/best.pt"
model = YOLO(MODEL_PATH)

def detect_image(image):
    results = model(image)
    annotated = results[0].plot()
    return annotated

def detect_video(video_path):
    cap = cv2.VideoCapture(video_path)

    # معلومات الفيديو
    width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps    = cap.get(cv2.CAP_PROP_FPS)

    output_path = "/content/output_detected.mp4"
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        results = model(frame)
        annotated = results[0].plot()
        out.write(annotated)

    cap.release()
    out.release()

    return output_path

with gr.Blocks(title="Vehicle Detection - YOLOv8") as demo:
    gr.Markdown("## 🚗 Vehicle Detection using YOLOv8")

    with gr.Tab("Image"):
        img_in = gr.Image(type="pil", label="Upload Image")
        img_out = gr.Image(label="Detected Image")
        btn_img = gr.Button("Detect")
        btn_img.click(detect_image, inputs=img_in, outputs=img_out)

    with gr.Tab("Video"):
        vid_in = gr.Video(label="Upload Video")
        vid_out = gr.Video(label="Detected Video")
        btn_vid = gr.Button("Detect")
        btn_vid.click(detect_video, inputs=vid_in, outputs=vid_out)

demo.launch(share=True)
```
